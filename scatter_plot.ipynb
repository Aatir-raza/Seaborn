```python
# scatterplot 
import matplotlib.pyplot as plt 
import seaborn as sns 

tips=sns.load_dataset('tips')
tips
```

```python
# scatter plot axes level function

sns.scatterplot(data=tips,x='total_bill',y='tip')

# relplot -> figure level (data=tips,x='total_bill",y='tip' , kind ='scatter')
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    kind='scatter'
)
```

```python
# scatter plot
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='sex',
    style='time',
    size='size'
)
```
