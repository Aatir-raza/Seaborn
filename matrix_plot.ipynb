```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Load dataset
tips = sns.load_dataset('tips')

# Select numerical columns
temp_df = tips.select_dtypes(include='number')

# Create figure
plt.figure(figsize=(15, 10))

# Create correlation heatmap
sns.heatmap(
    temp_df.corr(),
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

# Display plot
plt.title('Correlation Heatmap of Tips Dataset')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create sample Gapminder-style dataset
gap = pd.DataFrame({
    'country': ['India', 'India', 'India',
                'Germany', 'Germany', 'Germany',
                'France', 'France', 'France'],
    'continent': ['Asia', 'Asia', 'Asia',
                  'Europe', 'Europe', 'Europe',
                  'Europe', 'Europe', 'Europe'],
    'year': [2000, 2005, 2010] * 3,
    'lifeExp': [62, 64, 66,
                78, 79, 80,
                79, 80, 81]
})

# Filter European countries and create pivot table
temp_df = gap[gap['continent'] == 'Europe'].pivot(
    index='country',
    columns='year',
    values='lifeExp'
)

# Plot heatmap
plt.figure(figsize=(10, 5))
sns.heatmap(temp_df, annot=True, cmap='coolwarm', fmt='.1f')

plt.title('Life Expectancy in Europe')
plt.xlabel('Year')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```
